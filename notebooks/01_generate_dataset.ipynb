```python
import numpy as np
import pandas as pd

N = 100_000

df = pd.DataFrame({
    "customer_id": [f"C{i:06d}" for i in range(1, N + 1)]
})

df.head()
```

```python
print("Number of customers:", len(df))
```

```python
df.tail()
```

```python
df["age"] = np.random.randint(18, 71, size=N)

df[["customer_id", "age"]].head()
```

```python
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())
```

```python
df["past_purchases"] = np.random.poisson(
    lam=3,
    size=N
)

# Keep the value between 0 and 15
df["past_purchases"] = df["past_purchases"].clip(0, 15)

df[["customer_id", "age", "past_purchases"]].head()
```

```python
print("Minimum:", df["past_purchases"].min())
print("Maximum:", df["past_purchases"].max())
print("Average:", round(df["past_purchases"].mean(), 2))
```

```python
df["website_visits"] = np.random.poisson(
    lam=8,
    size=N
)

df["website_visits"] = df["website_visits"].clip(0, 40)
```

```python
df["product_views"] = np.random.poisson(
    lam=12,
    size=N
)

df["product_views"] = df["product_views"].clip(0, 60)

df[["customer_id", "website_visits", "product_views"]].head()
```

```python
df["avg_order_value"] = np.random.normal(
    loc=1500,
    scale=500,
    size=N
).round(2)

df["avg_order_value"] = df["avg_order_value"].clip(200, 5000)

df[["customer_id", "past_purchases", "avg_order_value"]].head()
```

```python
print("Minimum:", df["avg_order_value"].min())
print("Maximum:", df["avg_order_value"].max())
print("Average:", round(df["avg_order_value"].mean(), 2))
```

```python
df["days_since_last_purchase"] = np.random.randint(
    1, 181,
    size=N
)

df[["customer_id", "past_purchases", "days_since_last_purchase"]].head()
```

```python
print("Minimum:", df["days_since_last_purchase"].min())
print("Maximum:", df["days_since_last_purchase"].max())
print("Average:", round(df["days_since_last_purchase"].mean(), 2))
```

```python
df["loyalty_score"] = np.random.randint(
    1, 101,
    size=N
)

df[["customer_id", "past_purchases", "loyalty_score"]].head()
```

```python
print("Minimum:", df["loyalty_score"].min())
print("Maximum:", df["loyalty_score"].max())
print("Average:", round(df["loyalty_score"].mean(), 2))
```

```python
df["cart_additions"] = np.random.poisson(
    lam=3,
    size=N
)

df["cart_additions"] = df["cart_additions"].clip(0, 20)

df[["customer_id", "product_views", "cart_additions"]].head()
```

```python
print("Minimum:", df["cart_additions"].min())
print("Maximum:", df["cart_additions"].max())
print("Average:", round(df["cart_additions"].mean(), 2))
```

```python
df["email_engagement"] = np.random.randint(
    0, 101,
    size=N
)

df[["customer_id", "email_engagement"]].head()
```

```python
print("Minimum:", df["email_engagement"].min())
print("Maximum:", df["email_engagement"].max())
print("Average:", round(df["email_engagement"].mean(), 2))
```

```python
df["customer_tenure"] = np.random.randint(
    1, 61,
    size=N
)

df[["customer_id", "customer_tenure"]].head()
```

```python
print("Minimum:", df["customer_tenure"].min())
print("Maximum:", df["customer_tenure"].max())
print("Average:", round(df["customer_tenure"].mean(), 2))
```

```python
df["product_interest_score"] = np.random.randint(
    1, 101,
    size=N
)

df[["customer_id", "product_views", "cart_additions", "product_interest_score"]].head()
```

```python
print("Minimum:", df["product_interest_score"].min())
print("Maximum:", df["product_interest_score"].max())
print("Average:", round(df["product_interest_score"].mean(), 2))
```

```python
df["previous_discount_usage"] = np.random.poisson(
    lam=2,
    size=N
)

df["previous_discount_usage"] = df["previous_discount_usage"].clip(0, 10)

df[["customer_id", "past_purchases", "previous_discount_usage"]].head()
```

```python
print("Minimum:", df["previous_discount_usage"].min())
print("Maximum:", df["previous_discount_usage"].max())
print("Average:", round(df["previous_discount_usage"].mean(), 2))
```

```python
df["discount"] = np.random.choice(
    [0, 10, 20],
    size=N,
    p=[0.4, 0.3, 0.3]
)

df[["customer_id", "discount"]].head()
```

```python
print(df["discount"].value_counts().sort_index())
```

```python
baseline_score = (
    -3
    + 0.08 * df["loyalty_score"]
    + 0.10 * df["product_interest_score"]
    + 0.12 * df["past_purchases"]
    + 0.04 * df["website_visits"]
    + 0.03 * df["cart_additions"]
    - 0.008 * df["days_since_last_purchase"]
)
```

```python
baseline_probability = 1 / (1 + np.exp(-baseline_score))
```

```python
baseline_probability.head()
```

```python
treatment_effect = (
    0.002 * df["product_interest_score"]
    + 0.001 * df["loyalty_score"]
    + 0.003 * df["cart_additions"]
)
```

```python
discount_effect = (
    treatment_effect * (df["discount"] / 10)
)
```

```python
purchase_probability = baseline_probability + discount_effect
```

```python
purchase_probability = purchase_probability.clip(0.01, 0.99)
```

```python
print(purchase_probability.describe())
```

```python
df["purchase"] = np.random.binomial(
    1,
    purchase_probability,
    size=N
)

df[["customer_id", "discount", "purchase"]].head()
```

```python
print("Overall purchase rate:", round(df["purchase"].mean() * 100, 2), "%")
```

```python
print(
    df.groupby("discount")["purchase"]
      .mean()
      .mul(100)
      .round(2)
)
```

```python
Q1 = df_clean["avg_order_value"].quantile(0.25)
Q3 = df_clean["avg_order_value"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean["avg_order_value"] < lower_bound) |
    (df_clean["avg_order_value"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))
```
